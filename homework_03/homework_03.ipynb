## Задание 1: Конвертер регистров

Написать функцию, которая будет переводить snake_case в PascalCase и наоборот. 

Функция должна сама определять - какой формат ей передали. Можно добавить ключевой аргумент, который будет принудительно возвращать один из форматов.

<br>

**Примеры:**
* `otus_course     -> OtusCourse`
* `PythonIsTheBest -> python_is_the_best`



```python
# ВАШ КОД ЗДЕСЬ
import re
def transform(col):
    is_snake = "_" in col or col.islower()
    if is_snake: type = 'pascal'
    else: type = 'snake'
    if is_snake:
        words = col.split("_")
    else:
        col = re.sub(r"([A-Z]+)([A-Z][a-z])", r"\1_\2", col)
        col = re.sub(r"([a-z0-9])([A-Z])", r"\1_\2", col)
        words = col.split("_")
    if type == "snake":
        return "_".join(word.lower() for word in words)

    return "".join(word.capitalize() for word in words)
    
result = transform("SixSeven")
print(result)
```

## Задание 2: Проверка валидности даты

Написать функцию проверяющую валидность введенной даты.

<br>

**Примеры:**
* `29.02.2000 -> True`
* `29.02.2001 -> False`
* `31.04.1962 -> False`



```python
# ВАШ КОД ЗДЕСЬ
import re
from datetime import datetime


def valid_date(date_text):
    if re.fullmatch(r"[0-9]{2}\.[0-9]{2}\.[0-9]{4}", date_text) is None:
        return False

    try:
        datetime.strptime(date_text, "%d.%m.%Y")
    except ValueError:
        return False

    return True


date_text = input("Введите дату в формате ДД.ММ.ГГГГ: ")
print(valid_date(date_text))
```

## Задание 3: Проверка на простое число

Функция проверки на простое число. Простые числа – это такие числа, которые делятся на себя и на единицу.

<br>

**Примеры:**
* `17 -> True`
* `20 -> False`
* `23 -> True`

```python
# ВАШ КОД ЗДЕСЬ
def is_prime(number):
    if number < 2:
        return False

    divisor = 2
    while divisor * divisor <= number:
        if number % divisor == 0:
            return False
        divisor += 1

    return True


number = int(input("Введите целое число: "))
print(is_prime(number))
```

## Задание 4: Учет пользователей

Пользователь в бесконечном цикле вводит данные пользователей: имя, затем фамилию, возраст и ID. Ввод продолжается до тех пор, пока не будет введено пустое поле. 

Пользователи заносятся в словарь, где ключ это ID пользователя, а остальные данные записываются в виде кортежа. 

**Программа должна проверять:**
* имя и фамилия состоят только из символов и начинаются с большой буквы - если не с большой, то заменяет букву на большую;
* возраст должен быть числом от 18 до 60;
* ID - целое число, дополненное до 8 знаков незначащими нулями, ID должен быть уникальным.

**Дополнительно:** написать функцию, которая будет выводить полученный словарь в виде таблицы.

```python
# ВАШ КОД ЗДЕСЬ
def read_field(prompt, is_valid, error_message):
    while True:
        value = input(prompt).strip()
        if not value or is_valid(value):
            return value
        print(error_message)


def is_valid_age(value):
    if not value.isdecimal():
        return False

    try:
        return 18 <= int(value) <= 60
    except ValueError:
        return False


def print_users(users):
    rows = [("ID", "Имя", "Фамилия", "Возраст")]

    for user_id, (first_name, last_name, age) in users.items():
        rows.append((user_id, first_name, last_name, str(age)))

    widths = [max(len(value) for value in column) for column in zip(*rows)]

    for row in rows:
        print(" | ".join(value.ljust(width) for value, width in zip(row, widths)))


def main():
    users = {}

    def is_valid_id(value):
        return (
            value.isascii()
            and value.isdecimal()
            and len(value) <= 8
            and value.zfill(8) not in users
        )

    while True:
        first_name = read_field("Имя: ", str.isalpha, "Имя должно состоять из букв.")
        if not first_name:
            break

        last_name = read_field("Фамилия: ", str.isalpha, "Фамилия должна состоять из букв.")
        if not last_name:
            break

        age = read_field("Возраст: ", is_valid_age, "Введите целое число от 18 до 60.")
        if not age:
            break

        user_id = read_field("ID: ", is_valid_id, "Введите уникальный ID из 1–8 цифр.")
        if not user_id:
            break

        first_name = first_name[0].upper() + first_name[1:]
        last_name = last_name[0].upper() + last_name[1:]
        users[user_id.zfill(8)] = (first_name, last_name, int(age))

    print_users(users)


if __name__ == "__main__":
    main()
```
